```python
data = pd.read_csv("../dataset/Spotify_Winsorizado.csv")

df = data.copy()
```

```python
df['duration_ms']
```

```python
df["duration_min"] = df["duration_ms"] / 60000

df['duration_min']
```

```python
# Solo aplica para histogramas

print('Cantidad de bins para "duration_min" utilizando el método')
print(f'Freedman Diaconis: {freedman_diaconis_bins(df, 'duration_min')}')
```

```python
# Solo aplica para histogramas

print('Cantidad de bins para "duration_min" utilizando el método')
print(f'Sturges: {sturges_bins(df['duration_min'])}')
```

```python
df['duration_min']
```

```python
df['duration_category'] = pd.cut(
    df["duration_min"],
    bins=[0, 2.5, 4, 6, float("inf")],
    labels=["Corta", "Media", "Larga", "Muy larga"]
)

df.describe(include='category')
```

```python
df.drop(columns=["duration_ms"], inplace=True)
```
